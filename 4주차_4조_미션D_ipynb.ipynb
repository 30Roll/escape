#### 3. 3D 탐색을 통해 발견한 이상 징후 및 4D DEBUG 필요성

1. **정제 전 집계 결과의 뚜렷한 위화감 (동대문구 12배 착시)**

   * **원인 및 문제점:** 코드 실행상의 에러는 없으나, 데이터 내부의 **중복 수집 레코드(260건)** 누적 및 입력 오류로 인해 통계 왜곡이 발생함.

2. **수치형 컬럼(`생활인구수`)의 `object` 타입 인식 및 결측 형태 존재**
   * **관측 결과:** `생활인구수` 컬럼에 결측치 `NaN`(95건)과 문자열 `'-'`(100건)이 혼재되어 수치 연산이 불가능한 상태임.
   * **원인 및 문제점:** 정제 없이 연산 시 `TypeError`가 발생하므로 `pd.to_numeric()` 변환 및 결측 처리가 시급함.

---

* **3D 결론:** "코드는 에러 없이 정상 작동했지만 답은 틀렸습니다." 1차 집계에서 포착한 위화감을 해결하기 위해 **4D DEBUG(5대 데이터 정제 표준 순서)** 작업이 필수적으로 수행되어야 함을 입증함.
---
>  **AI 활용 및 검증 기록 (3D)**
> - **질문**: 3D 탐색 실행 결과 표 가공 요청
> - **수신**: 1차 집계 결과(동대문구 12배 이슈 포함) 및 마크다운 표 구조 수령
> - **검증**: 코랩 실행 후 정제 전 동대문구 인구(1,231만 명)의 통계적 위화감 직접 확인
> - **결과**: **채택** (4D 정제 단계의 필요성 입증)

```python
# ==========================================
# [3D DISCOVER] 첫인상 5줄 표준 탐색 및 정제 전 1차 집계
# ==========================================
import pandas as pd
import numpy as np
import io
from google.colab import files
from IPython.display import display

# ------------------------------------------
# 0. 파일 업로드 및 기초 전처리
# ------------------------------------------
print(" '실습데이터_서울시_생활인구_2026.csv' 파일을 선택하여 업로드해주세요.")
uploaded = files.upload()

file_name = list(uploaded.keys())[0]
df = pd.read_csv(io.BytesIO(uploaded[file_name]), encoding='cp949')

# 컬럼명 앞뒤 불필요 공백 제거 (KeyError 예방)
df.columns = df.columns.str.strip()

# ------------------------------------------
# 1. 첫인상 5줄 표준 탐색 루틴
# ------------------------------------------
print("\n" + "=" * 65)
print(" [3D DISCOVER] 첫인상 5줄 표준 탐색 루틴")
print("=" * 65)

# ① df.shape (데이터 규모)
print(f"① [df.shape] 데이터 규모 (행, 열) : 총 {df.shape[0]:,}행 × {df.shape[1]}열")

# ② df.head(3) (상위 3행 샘플)
print("\n② [df.head(3)] 상위 3개 행 샘플:")
display(df.head(3))

# ③ df.dtypes & ⑤ df.isna().sum() & 특수기호('-') 통합 정리
info_df = pd.DataFrame({
    '데이터 타입': df.dtypes.astype(str),
    '결측치(NaN)': df.isna().sum(),
    '특수기호(\'-\')': (df == '-').sum()
})
print("\n③&⑤ [컬럼별 데이터 상태 통합 요약 (타입, 결측치, 특수기호)]")
display(info_df)

# ④ df.describe() (요약 통계량)
print("\n④ [df.describe()] 주요 요약 통계량 (수치 및 범주):")
display(df.describe(include='all'))

# ------------------------------------------
# 2. 정제 전 1차 집계 및 위화감 포착
# ------------------------------------------
print("\n" + "=" * 65)
print(" [3D DISCOVER] 정제 전 자치구별 생활인구 합계 TOP 5")
print("=" * 65)

# 정제 전 단순 합계 집계 (문자열 등 숫자 불가능 값은 NaN 처리 후 합산)
raw_pop = pd.to_numeric(df['생활인구수'], errors='coerce')
top5_before = (df.assign(v=raw_pop)
                .groupby('자치구')['v'].sum()
                .sort_values(ascending=False)
                .head(5)
                .reset_index())

# 보기 쉽게 천 단위 쉼표(,) 및 '명' 붙이기
top5_before.columns = ['자치구', '정제 전 생활인구 합계']
top5_before['정제 전 생활인구 합계'] = top5_before['정제 전 생활인구 합계'].apply(lambda x: f"{int(x):,}명")

display(top5_before)

print("=" * 65)
print(" [위화감 포착]: 1위 동대문구가 2위 강남구보다 12배 이상 큼!")
print("    문법 에러나 결측치 오류는 없으나, 중복 입력 등으로 답이 틀렸음을 입증 (4D DEBUG 필수)")
print("=" * 65)
```

### 4D DEBUG (데이터 정제 및 품질 검증)

* **작성자:** 고륜
* **작업 개요:** 3D DISCOVER 단계에서 포착된 통계적 위화감(동대문구 새벽 시간대 12배 인구 착시 현상)을 해결하기 위해 5대 정제 표준 순서에 따라 데이터 청정화 및 품질 검증 수행.
* **핵심 정제 원칙 :**
  1. **무음 실패 사전 차단**: `시간대` 컬럼의 `int64` 정수 타입 특성을 고려하여 `isin([2, 3, 4])` 정수 필터링을 적용하여 필터링 조건 불일치로 인한 0건 정제 오류 방지.
  2. **IQR 기계적 삭제 오답 방지**: 강남구 등 실제 인구가 많은 고밀도 구역의 데이터를 기계적으로 삭제하는 오류를 방지하고, 물리적으로 불가능한 이상치만 정밀 타겟 삭제.
  3. **미션 범위 행 수(Non-Zero) 실시간 검증**: 각 정제 단계마다 분석의 핵심 대상인 **방과후 시간대(15~18시) 10대 데이터**가 0행으로 증발하지 않고 유지되는지 `check_mission_rows()` 함수 및 `assert` 구문으로 검증.

```python
# ==========================================
# [4D DEBUG] 5대 표준 순서 및 미션 범위 행 수(Non-Zero) 검증 코드
# ==========================================
import pandas as pd
import numpy as np
from IPython.display import display

#  [수정] 파일 로드 구문 대신 기존 df를 복사하여 사용합니다.
# (이전 단계에서 불러온 df 변수를 그대로 사용)
df_clean = df.copy()
initial_len = len(df_clean)

# ------------------------------------------
# 0. 미션 범위 (15~18시, 10대) 행 수 검증 함수 정의
# ------------------------------------------
def check_mission_rows(dataframe, step_name):
    """정제 각 단계 후 미션 범위 데이터가 0행이 아닌지(Non-Zero) 실시간 검증"""
    target = dataframe[
        (dataframe['시간대'] >= 15) &
        (dataframe['시간대'] <= 18) &
        (dataframe['연령대'].astype(str).str.contains('10', na=False))
    ]
    count = len(target)
    assert count > 0, f" [CRITICAL ERROR] '{step_name}' 후 미션 타겟(15~18시, 10대) 데이터가 0행이 되었습니다! 필터링 조건을 점검하세요."
    return count

print("=" * 70)
print(" [4D DEBUG] 5대 데이터 정제 표준 순서 적용 및 정제 로그 작성")
print("=" * 70)


# ① 타입 정리 및 무음 실패(Silent Failure) 예방
df_clean.columns = df_clean.columns.str.strip()
df_clean['자치구'] = df_clean['자치구'].astype(str).str.strip()
df_clean['연령대'] = df_clean['연령대'].astype(str).str.strip()

# '시간대'를 정수형(int64)으로 변환하여 무음 실패 예방
df_clean['시간대'] = pd.to_numeric(df_clean['시간대'], errors='coerce').astype('Int64')

# '-' 특수기호를 NaN으로 변환 후 수치형 타입 변경
df_clean['생활인구수'] = pd.to_numeric(df_clean['생활인구수'].replace('-', np.nan), errors='coerce')

mission_cnt = check_mission_rows(df_clean, "1단계: 타입 정리")
print(f"① [1단계: 타입 정리] '시간대'(int64) 및 '생활인구수' 수치형 변환 완료 (전체 {initial_len:,}행 / 미션 범위: {mission_cnt:,}행)")


# ② 중복 데이터 처리
len_before_dup = len(df_clean)
df_clean = df_clean.drop_duplicates()
dup_removed = len_before_dup - len(df_clean)

mission_cnt = check_mission_rows(df_clean, "2단계: 중복 제거")
print(f"② [2단계: 중복 제거] [처리] 중복 레코드 제거 / [건수] {dup_removed:,}건 / [영향] 잔여 {len(df_clean):,}행 (미션 범위: {mission_cnt:,}행)")


# ③ 결측치 처리
len_before_null = len(df_clean)
df_clean = df_clean.dropna(subset=['자치구', '시간대', '연령대', '생활인구수'])
null_removed = len_before_null - len(df_clean)

mission_cnt = check_mission_rows(df_clean, "3단계: 결측치 제거")
print(f"③ [3단계: 결측 처리] [처리] 결측치('-' 및 NaN) 제거 / [건수] {null_removed:,}건 / [영향] 잔여 {len(df_clean):,}행 (미션 범위: {mission_cnt:,}행)")


# ④ 논리적/통계적 이상치 타겟 처리
len_before_outlier = len(df_clean)
bad_dongdaemun = (df_clean['자치구'] == '동대문구') & (df_clean['시간대'].isin([2, 3, 4]))
negative_values = df_clean['생활인구수'] < 0

df_clean = df_clean[~(bad_dongdaemun | negative_values)].copy()
outlier_removed = len_before_outlier - len(df_clean)

mission_cnt = check_mission_rows(df_clean, "4단계: 이상치 제거")
print(f"④ [4단계: 이상치 정제] [처리] 동대문구 02~04시 이상치 제거 / [건수] {outlier_removed:,}건 / [영향] 잔여 {len(df_clean):,}행 (미션 범위: {mission_cnt:,}행)")


# ⑤ 강남구 등 정상 고인구 보존
gangnam_count = (df_clean['자치구'] == '강남구').sum()
print(f"⑤ [5단계: 고인구 보존] [처리] 강남구 고인구 보존 / [건수] 0건 삭제 ({gangnam_count:,}행 보존)")

# 정제 완료 데이터 업데이트
df = df_clean.copy()


# ------------------------------------------
# [최종 무음 실패 검증] 미션 범위(15~18시, 10대) 종합 평가
# ------------------------------------------
print("\n" + "=" * 70)
print(" [최종 무음 실패 검증] 미션 타겟 범위 (15~18시, 10대) 행 수 기준 확인")
print("=" * 70)

final_mission_rows = check_mission_rows(df, "최종 검증")
print(f" [미션 타겟 최종 행 수]: 총 {final_mission_rows:,}행 확인")
print(" [검증 기준 통과]: 미션 타겟 데이터가 0행이 아닌 정상 상태(Non-Zero)로 살아있어 5D 분석 진행이 가능합니다!")


# ------------------------------------------
# 정제 후 전체 자치구 재집계 TOP 5 및 최종 행 수 출력
# ------------------------------------------
print("\n" + "=" * 70)
print(" [4D DEBUG] 정제 후 자치구별 생활인구 합계 TOP 5 (재집계)")
print("=" * 70)

top5_after = (df.groupby('자치구')['생활인구수'].sum()
                .sort_values(ascending=False)
                .head(5)
                .reset_index())

top5_after.columns = ['자치구', '정제 후 생활인구 합계']
top5_after['정제 후 생활인구 합계'] = top5_after['정제 후 생활인구 합계'].apply(lambda x: f"{int(x):,}명")

display(top5_after)

# 정제 후 최종 행 수 요약 출력
print("-" * 70)
print(f" [정제 결과 요약] 원본 데이터: {initial_len:,}행 -> 정제 후 최종 클린 데이터: {len(df):,}행")
print(f" [미션 대상 요약] 정제 후 15~18시 10대 데이터: {final_mission_rows:,}행 보존 완료")
print("=" * 70)
```

```python
# ==========================================
# [5D DELIVER] 15~18시 10대 생활인구 분석 및 표 도출
# ==========================================
import pandas as pd
from IPython.display import display

# 1. 미션 대상 조건 필터링 (15~18시, 10대)
df_15_18 = df[(df['시간대'] >= 15) & (df['시간대'] <= 18)].copy()
df_10s = df_15_18[df_15_18['연령대'].astype(str).str.contains('10', na=False)]

# 2. 자치구별 10대 절대인구 및 전연령 인구 집계
abs_df = df_10s.groupby('자치구')['생활인구수'].sum().reset_index(name='10대_절대인구')
total_df = df_15_18.groupby('자치구')['생활인구수'].sum().reset_index(name='전연령_인구')

# 3. 데이터 병합 및 수치형 타입 안전 보장
result_df = pd.merge(abs_df, total_df, on='자치구')
result_df['10대_절대인구'] = pd.to_numeric(result_df['10대_절대인구'])
result_df['전연령_인구'] = pd.to_numeric(result_df['전연령_인구'])

# 4. 10대 비중(%) 및 교차 순위 산출
result_df['10대_비중(%)'] = (result_df['10대_절대인구'] / result_df['전연령_인구']) * 100
result_df['절대순위'] = result_df['10대_절대인구'].rank(ascending=False, method='min').astype(int)
result_df['비중순위'] = result_df['10대_비중(%)'].rank(ascending=False, method='min').astype(int)
result_df['순위합계'] = result_df['절대순위'] + result_df['비중순위']

# 5. 출력용 포맷팅 ('순위합계' 우수순 정렬)
display_df = result_df.sort_values(by='순위합계', ascending=True).copy()

display_df['10대_절대인구'] = display_df['10대_절대인구'].apply(lambda x: f"{int(x):,}명")
display_df['전연령_인구'] = display_df['전연령_인구'].apply(lambda x: f"{int(x):,}명")
display_df['10대_비중(%)'] = display_df['10대_비중(%)'].apply(lambda x: f"{x:.2f}%")

# 6. 최종 교차 분석표 및 요약 브리핑 출력
print("============================================================")
print(" [5D DELIVER] 15~18시 자치구별 10대 생활인구 교차 분석표 (종합 순위순)")
print("============================================================")
display(display_df[['자치구', '10대_절대인구', '전연령_인구', '10대_비중(%)', '절대순위', '비중순위', '순위합계']])

top3 = display_df.head(3)['자치구'].tolist()
print("\n" + "=" * 60)
print(" [분석 요약] 교차 순위(절대순위+비중순위) 종합 최우선 추천 자치구 Top 3:")
print(f"   1위: {top3[0]} | 2위: {top3[1]} | 3위: {top3[2]}")
print("============================================================")
```

#### 1. 15~18시 자치구별 10대 생활인구 교차 분석표 (상위 10개 자치구)
